# Overview of database and generated results

## 1. Results
Visualize the Prediction results (-> result.pkl) which is generated in eval_utils:

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from tabulate import tabulate

# Load the result.pkl file
result_file_path = 'output/models/kitti/VirConv-T/Subset_Eight_Hundred/eval/epoch_40/val/default/result.pkl'

with open(result_file_path, 'rb') as f:
    result = pickle.load(f)
    # print(result)


# Create a DataFrame from the result
data = []
for item in result:
    for i in range(len(item['name'])):
        data.append({
            'name': item['name'][i],
            'truncated': item['truncated'][i],
            'occluded': item['occluded'][i],
            'alpha': item['alpha'][i],
            'bbox': item['bbox'][i],
            'dimensions': item['dimensions'][i],
            'location': item['location'][i],
            'rotation_y': item['rotation_y'][i],
            'score': item['score'][i],
            'frame_id': item['frame_id']
        })

df = pd.DataFrame(data)

# Print the DataFrame as a table
print(tabulate(df, headers='keys', tablefmt='psql'))
```

**kitti_dbinfos_train_mm:**

Contains:
- name	
- path	
- image_idx	
- gt_idx	
- box3d_lidar	(7 values)
- num_points_in_gt

```python
import pickle
import pandas as pd

# Load the .pkl file
file_path = "/home/leo/workspace/Docker_tests/Torch1131CUDA117/VirConv/data/kitti/kitti_dbinfos_train_mm.pkl" 
with open(file_path, "rb") as f:
    data = pickle.load(f)

# Check if the data is a dictionary
if isinstance(data, dict):
    print("Keys in the dataset:", list(data.keys()))

    # Select a sample key and display its structure
    sample_key = list(data.keys())[0]
    sample_value = data[sample_key]

    if isinstance(sample_value, list) and len(sample_value) > 0:
        # Convert first few entries into a DataFrame for better visualization
        df = pd.DataFrame(sample_value) 
        pd.set_option("display.max_columns", None)  # Show all columns
        # pd.set_option("display.max_rows", 100)  # Adjust row display limit
        pd.set_option("display.max_colwidth", None)  # Show full content in columns

        display(df)  # Works in Jupyter Notebook
    else:
        print("Data under this key is not in list format.")
else:
    print("Data is not a dictionary.")

# Print Examples values for one image_index
print("Selected image index: ", sample_value[0]['image_idx'])
print(f"Lidar point cloud data path: {sample_value[0]['path']}")

# output the number of detected object for the first image_idx 
image_idx = sample_value[0]['image_idx']
num_objects = sum(1 for item in sample_value if item['image_idx'] == image_idx)
print("Number of objects in the image index: ", num_objects)

print("\nExample for the Box3D Lidar: \n")
print(sample_value[0]['box3d_lidar'])

```
